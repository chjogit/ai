- 중심위치 측도 : 평균, 중앙값, 최빈값
- 산포 측도 : 왜도, 첨도

```python
import numpy as np
from scipy.stats import skew, kurtosis # 왜도0, 첨도0이면 정규분포
```

```python
# np.random.normal(loc=0.0, scale=1.0, size=tup) : 평균 loc, 표준편차 scale을 갖는 정규분포 랜덤값
data = np.random.normal(0, 1, size=10000000)
```

```python
print('평균 :', np.mean(data), data.mean())
print('중위수 :', np.median(data))
print('분산 :', np.var(data), data.var())
print('표준편차 :', np.std(data))
```

```python
print('왜도 :', skew(data)) # 왜도<0 : 왼쪽 긴꼬리 데이터(오른쪽으로 치우친 데이터)
print('첨도 :', kurtosis(data)) # 양수면 정규분포보다 뽀족한 봉우리 모양의 분포
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(5,1))
plt.hist(data, bins=100)
plt.show()
```

```python
from scipy.stats import skewnorm, norm
# a<0:왼쪽꼬리긴데이터(왜도는 아님)
data = skewnorm(a=-5, loc=0, scale=1).rvs(size=10000000)
print('평균 :', np.mean(data))
print('중위수 :', np.median(data))
print('분산 :', np.var(data))
print('표준편차 :', np.std(data))
print('왜도 :', skew(data))
print('첨도 :', kurtosis(data))
plt.figure(figsize=(5,1))
plt.hist(data, bins=300)
plt.show()
```

```python
from sklearn.preprocessing import (StandardScaler, MinMaxScaler, MaxAbsScaler, 
                                   RobustScaler, Normalizer)
df = pd.DataFrame({'x1':range(11),
                   'x2':[-100, -80, -60, -40, -20, 0, 20, 40, 60, 80, 500]})
df
```

```python
# StandardScaler : 평균0, 표준편차1(표준화)
scaler = StandardScaler()
scaler.fit(df)
ary_std = scaler.transform(df)
df_std = pd.DataFrame(ary_std, columns=['x1_std', 'x2_std'])
result = pd.concat([df, df_std], axis=1)
result
```

```python
# MinMaxScaler : 0~1사이(정규화)
scaler = MinMaxScaler()
ary_min = scaler.fit_transform(df)
df_min = pd.DataFrame(ary_min, columns=['x1_min', 'x2_min'])
result = pd.concat([result, df_min], axis=1)
result
```

```python
# MaxAbsScaler : -1~1사이
scaler = MaxAbsScaler()
ary_max = scaler.fit_transform(df)
df_max = pd.DataFrame(ary_max, columns=['x1_max', 'x2_max'])
result = pd.concat([result, df_max], axis=1)
result
```

```python
# RobustScaler : 중앙값, 사분위수, IQR(Q3-Q1)기준으로 변환
# 극단값들에 대한 영향을 적게 받는 특징 있음
scaler = RobustScaler()
ary_rob = scaler.fit_transform(df)
df_rob = pd.DataFrame(ary_rob, columns=['x1_rob', 'x2_rob'])
result = pd.concat([result, df_rob], axis=1)
result
```

```python
# Normalizer : 군집분석
scaler = Normalizer()
ary_nor = scaler.fit_transform(df)
df_nor  = pd.DataFrame(ary_nor, columns=['x1_nor', 'x2_nor'])
result = pd.concat([result, df_nor], axis=1)
result
```

### 평균값 인코딩
- 레이블 값을 수치적으로 표현하하면서도 서로 구분할 수 있음
- 장점 : 모델의 비교적 빠른 학습함. 회귀분석 뿐만 아니라, 분류분석에서도 예측값에 좀 더 가깝게 학습됨
- 단점 : 과적합

```python
from seaborn import load_dataset
titanic = load_dataset('titanic')
# male과 female => 0, 1 (레이블 인코딩)
# male과 female => 1 0, 0 1 (원핫인코딩)
# male과 femail => 남자평균생존률, 여자평균생존율 (평균값인코딩)
# 성별 생존율
sex_mean = titanic.groupby('sex')['survived'].mean()
sex_mean
```

```python
titanic['sex_mean'] = titanic['sex'].map(sex_mean) # 맨 끝 열에  sex_mean 추가
titanic.drop('sex_mean', axis=1, inplace=True)
```

```python
# 3번째 열에 sex_mean 추가
titanic.insert(3, 'sex_mean', titanic['sex'].map(sex_mean))
titanic.head()
```

```python
# 위처럼 하나의 열로 만들지 말고, 원핫인코딩처럼 두개의 열로 쓰자
titanic_pivot = titanic.pivot_table(columns='survived',
                   index=titanic.index,
                   values='sex_mean')
titanic_pivot.fillna(0)
```

## 3.3 결측값 처리(대체)

```python
from seaborn import load_dataset
iris = load_dataset('iris')
iris_X = iris.iloc[:, :-1]
iris_y = iris.iloc[:, -1]
iris_X.isna().sum()
```
