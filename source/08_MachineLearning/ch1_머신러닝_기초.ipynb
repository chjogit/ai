## 2.1 변수의 종류

- 데이터 속성에 따른 4대 변수 분류 체계

    - 명목척도(명목식): 범주 구분을 위한 척도 (예: 남, 여)

    - 순서척도(순서식): 서열 관계가 존재하는 척도 (예: 상/중/하)

    - 구간척도(구간식): 순서 간 간격 분석의 의미를 갖는 간격 척도

    - 비율척도(비율식): 절대 영점 및 연속성을 보유한 척도

- 변수 타입(형태)에 따른 구분

    - 범주형, 이산형, 연속형 변수 유형

- 데이터 품질(상태)에 따른 구분

    - 특이값(이상치), 결측값 존재 유무에 따른 데이터 구분

## 2.2 통계 요약

- 통계적 요약 측도의 분류
    - 중심 위치 측도 : 데이터 분포의 대표 중심값을 나타내는 평균값, 중앙값, 최빈값(최반값)
    - 산포 측도 : 데이터 분포의 치우침 및 비대칭성을 나타내는 왜도, 첨도

````python
import numpy as np
from scipy.stats import skew, kurtosis # 왜도와 첨도가 모두 0인 상태 : 정규분포
````

````python
# 정규 분포 랜덤 값
data = np.random.normal(
    loc=0.0, # 평균
    scale=1.0, # 표준편차
    size=10000000 # 크기
)
````

````python
print('평균 :', np.mean(data))
print('중위수 :', np.median(data))
print('분산 :', np.var(data))
print('표준편차 :', np.std(data))
````

````python
print('왜도 :', skew(data))
print('첨도 :', kurtosis(data))
````

````python
import matplotlib.pyplot as plt
plt.figure(figsize=(5,1))
plt.hist(data, bins=100)
plt.show()
````

````python
from scipy.stats import skewnorm, norm

data = skewnorm(
    a=5, # 왜도(Skewness) 형상 파라미터 : 오른쪽으로 길게 늘어진 비대칭 분포(양의 왜도) 형성
    loc=0, # 중심 파라미터
    scale=1 # 분포의 척도(Scale)
).rvs(size=10000000)
````

````python
print('평균 :', np.mean(data))
print('중위수 :', np.median(data))
print('분산 :', np.var(data))
print('표준편차 :', np.std(data))
````

## 2.5 상관 분석
- 두 변수 간 선형적 연관성(관계) 존재 여부 규명을 위한 분석 기법

### 공분산과 상관계수

* **공분산(Covariance)**
    * 두 변수의 함께 변화하는 경향성 및 직선적 관련성 방향(양의 상관 / 음의 상관) 측정 도구
    * 측정 단위의 영향을 받아 크기 자체로 상관관계의 강도를 비교하기 어려운 한계성
    
* **상관관계(Correlation)**
    - corr() : 전체 변수 간 상관계수 행렬 출력
    - pearsonr() : 두 변수 간 피어슨 상관계수 및 통계적 유의성 검정을 위한 p-value 산출 
        - p-value : 귀무가설이 참이라는 가정 하에, 현재 관측된 결과만큼 극단적인 결과가 실제로 일어날 확률
---

* 귀무가설 ($H_0$)
    * 검정 대상이 되는 기본 가설로서, 기존의 입장, 변화 없음, 차이 없음, 효과 없음을 나타내는 가설
    * 검정 메커니즘 : 통계적 가설 검정 과정에서 직접적인 기각을 목적으로 수립되는 가설
    * 채택 및 기각 기준 : P-value가 유의수준($\alpha$)보다 작은 경우 기각되며, 기각 시 대립가설의 채택으로 연결되는 구조
        - 유의수준 : 귀무가설을 잘못 기각할 확률의 최대 허용 한계값 ; 5%, 1%, 10% 등 

* 대립가설 ($H_1$ 또는 $H_a$)
    * 연구자 또는 분석가가 입증하고자 하는 새로운 주장, 차이 있음, 효과 있음, 변화 있음을 나타내는 가설
    * 검정 메커니즘 : 귀무가설이 기각될 때 최종적으로 채택 및 지지되는 가설

---

* p-value가 높을 경우 (p-value $> \alpha$, 예: $p > 0.05$)
    * 귀무가설과의 관계 : 귀무가설을 기각할 수 없음 (귀무가설 채택/기각 실패)
    * 통계적 유의성 : 통계적으로 유의미한 차이나 효과가 존재하지 않음 (Not Statistically Significant)
    * 해석 및 의미
        * 관측된 데이터의 차이 및 연관성이 단순한 표본 오차나 우연에 의해 발생했을 가능성이 높음
        * 새로운 주장(대립가설)을 뒷받침할 만한 충분한 통계적 증거의 부족




* p-value가 낮을 경우 (p-value $\le \alpha$, 예: $p \le 0.05$)
    * 귀무가설과의 관계 : 귀무가설 기각 및 대립가설 채택
    * 통계적 유의성 : 통계적으로 유의미한 차이나 효과가 존재함 (Statistically Significant)
    * 해석 및 의미
        * 귀무가설하에서 현재와 같은 데이터가 우연히 관측될 확률이 극히 희박함
        * 연구자가 입증하고자 하는 새로운 주장이나 변수 간 연관성(대립가설)이 통계적으로 입증됨

---

```
모수적 방법 : 데이터가 정규분포와 같은 정형화된 수학적 분포 형태를 취하고 있다고 가정하기 때문에, 평균이나 표준편차 같은 '모수'를 사용해 가설 검정을 하거나 상관관계를 분석합니다.

비모수적 방법 : 데이터의 수가 적거나 정규분포를 따르지 않는 경우, 모수(평균·표준편차) 대신 순위나 중앙값 등을 활용하여 분석하는 방식입니다.
```

````python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler, RobustScaler, Normalizer
````

````python
df = pd.DataFrame({'x1':range(11),
                   'x2':[-100, -80, -60, -40, -20, 0, 20, 40, 60, 80, 500]})
df
````

````python
# StandardScaler : 평균 0, 표준편차 1
scaler = StandardScaler()
scaler.fit(df)
array_std = scaler.transform(df)
df_std = pd.DataFrame(array_std, columns=['x1_std','x2_std'])
result = pd.concat([df,df_std], axis=1)
````

````python
# MinMaxScaler : 0 ~ 1
scaler = MinMaxScaler()
scaler.fit(df)
array_min = scaler.transform(df)
df_min = pd.DataFrame(array_min, columns=['x1_min','x2_min'])
result = pd.concat([result, df_min], axis=1)
````

````python
# MaxAbsScaler : -1 ~ 1
scaler = MaxAbsScaler()
scaler.fit(df)
array_abs = scaler.transform(df)
df_abs = pd.DataFrame(array_abs, columns=['x1_abs','x2_abs'])
result = pd.concat([result, df_abs], axis=1)
````

````python
# RobustScaler : 중앙값 0, IQR 기준
scaler = RobustScaler()
scaler.fit(df)
array_rob = scaler.transform(df)
df_rob = pd.DataFrame(array_rob, columns=['x1_rob','x2_rob'])
result = pd.concat([result, df_rob], axis=1)
````

````python
# Normalizer : 벡터 크기 1
scaler = Normalizer()
scaler.fit(df)
array_nor = scaler.transform(df)
df_nor = pd.DataFrame(array_nor, columns=['x1_nor','x2_nor'])
result = pd.concat([result, df_nor], axis=1)
````

### 평균값 인코딩
- 각 범주가 타깃 레이블에 미치는 영향을 직접적인 수치로 표현하면서 범주 간 고유한 구분이 가능

````python
from seaborn import load_dataset
titanic = load_dataset('titanic')

# male, female → 레이블 인코딩 (0, 1), 원-핫 인코딩 (10, 01), 평균값 인코딩 (남자 평균 생존률, 여자 평균 생존률)
sex_mean = titanic.groupby('sex')['survived'].mean()
sex_mean
````

````python
titanic.insert(3, 'sex_mean', titanic['sex'].map(sex_mean))
````

````python
titanic.head()
````

````python
# 원-핫 인코딩처럼 두 개의 열로 표현하기
titanic_pivot = titanic.pivot_table(
    columns='survived',
    index=titanic.index,
    values='sex_mean'
)

titanic_pivot.fillna(0)
````

### 범주형 데이터 인코딩 기법 비교


* **레이블 인코딩 (Label Encoding)**
    * **장점** : 범주 개수와 무관한 단일 컬럼 유지로 메모리 효율성 확보 및 단순 정수 변환을 통한 빠른 연산 속도와 처리 간소화
    * **단점** : 임의의 수치 크기(0, 1, 2...) 부여에 따른 불필요한 순서/서열 관계 형성 및 선형/거리 기반 모델에서의 잘못된 가중치 반영 가능성
    * **적용 상황** : 순서형 데이터(예: 학점, 직급) 및 트리 기반 모델(Decision Tree, Random Forest, XGBoost 등) 적용 시


* **원-핫 인코딩 (One-Hot Encoding)**
    * **장점** : 이진 벡터화를 통한 범주 간 독립성 보장 및 불필요한 순서 관계와 수치적 편향의 근본적 차단
    * **단점** : 범주 수 증가 시 차원의 폭발적 확장(희소 행렬 형성) 및 메모리 소모 증가와 차원의 누적(Curse of Dimensionality)에 의한 과적합 위험
    * **적용 상황** : 명목형 데이터(예: 성별, 국적, 혈액형), 범주 종류(Unique Values)가 적은 데이터셋, 선형 모델, 신경망, 거리 기반 모델(KNN, SVM) 적용 시


* **평균값 인코딩 (Mean / Target Encoding)**
    * **장점** : 단일 수치 컬럼 변환을 통한 차원 확장 방지 및 타깃 변수의 통계적 특성 직접 반영을 통한 높은 예측 성능
    * **단점** : 학습 데이터 타깃 정보 반영에 따른 타깃 누수(Target Leakage) 및 과적합 위험과 Smoothing 및 Cross-Validation 미적용 시 테스트 성능 저하
    * **적용 상황** : 고수준 범주형 데이터(High-Cardinality, 예: 우편번호, 상품ID) 및 K-Fold 등을 적용한 고성능 머신러닝 모델 구축 시
