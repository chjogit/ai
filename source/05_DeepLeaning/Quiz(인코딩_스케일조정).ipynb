    * 표준화 작업 후 : 지역명2s, 연도s, 월s 컬럼으로 추가 (df_s)

```python
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Input
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.preprocessing import LabelEncoder
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('https://raw.githubusercontent.com/4aix/data/refs/heads/master/ch13_apt_fillna_median.csv')
df = pd.DataFrame(data)
```

# 지역명 라벨 인코딩

```python
df.get('지역명')
```

```python
labeled_data = LabelEncoder().fit_transform(df.get('지역명'))
```

```python
df['지역명2'] = labeled_data
```

```python
df
```

# MinMaxScaling

```python
X_data = df[['지역명2', '연도','월']]
y_data = df[['평당분양가격']]
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

norm_scaled_X = scaler_X.fit_transform(X_data)
norm_scaled_y = scaler_y.fit_transform(y_data)
```

```python
df_m = pd.DataFrame(norm_scaled_X, columns=['지역명2m', '연도m', '월m'])
df_m['평당분양가격m'] = norm_scaled_y
```

```python
df_m
```

# StandardScaling

```python
X_data = df[['지역명2', '연도','월']]
y_data = df[['평당분양가격']]
scaler_X = StandardScaler()
scaler_y = StandardScaler()

std_scaled_X = scaler_X.fit_transform(X_data)
std_scaled_y = scaler_y.fit_transform(y_data)

df_s = pd.DataFrame(std_scaled_X, columns=['지역명2s', '연도s', '월s'])
df_s['평당분양가격s'] = std_scaled_y
```

```python
df_s
```
