## 2.5 배열의 차원 변경
- reshape(행, 열): -1 사용 가능
    - 원본의 총 원소 개수와 바꿀 모양의 총 원소 개수가 반드시 일치해야함
    - 원본 데이터는 그대로 두고, 모양이 바뀐 "새로운 뷰(View)"를 반환함
- resize(tup): -1 사용 불가
    - 원본 배열 자체를 직접 변경
    - 원본과 원소 개수가 달라도 작동
- ravel() : 1차원 배열(View)을 반환, ravel()로 만든 배열의 값을 수정할 경우, 원본도 함께 바뀜
- flatten() : 1차원으로 변경한 새로운 복사본(Copy) 배열을 반환
- T, transpose() : 전치(행과 열을 바꿈)행렬로 View 반환

- 두 배열을 쌓아 합치는 함수
    * vstack(tup) : 아래로 추가하는 방식 (Vertical)
    * hstack(tup) : 옆으로 추가하는 방식 (Horizontal)
    * dstack(tup) : 깊이로 추가하는 방식 (Depth) ; 차원 증가

````python
c = np.array([1,2,3])
d = np.array([4,5,6])
v = np.vstack((c,d))
print('vstack - ', v, 'shape - ',v.shape, sep='\n')
h = np.hstack((c,d))
print('hstack - ', h, 'shape - ',h.shape, sep='\n')
d = np.dstack((c,d))
print('dstack - ', d, 'shape - ',d.shape, sep='\n')
````

````python
x = np.array([2,4]) # 독립 변수(입력 데이터)
y = np.array([40,60]) # 종속 변수(타겟 변수)
Y = y.reshape(-1,1)
X = np.c_[x, np.ones(len(x))]  # [2,4]와 [1,1]의 각 배열을 세로로 추가
````

````python
#     X⁻¹@ X = I
wb = np.linalg.inv(X)@Y
wb
````

````python
# 가중치 weight, 편향 bias / y = wx + b
wb[0], wb[1]
````

````python
# 도출된 회귀식 y = wx + b으로 계산된 예측값 h
x_ = np.linspace(0, 10, 300)
h = wb[0]*x_ + wb[1] 
h = np.add(np.multiply(wb[0], x_), wb[1]) # 데이터가 많고, 식이 복잡할 경우 범용함수 사용을 권장함
````

````python
import matplotlib.pyplot as plt
plt.figure(figsize=(4,3)) # 현재 셀의 그래프 사이즈 조정
plt.rcParams['figure.figsize'] = 4,3 # 그래프 사이즈 default 조정
plt.scatter(x_, h, c='r', s=1) # 점그래프 scatter(X, Y, color, size)
plt.scatter(x, y) # 실제 데이터
plt.show()
````

````python
# X가 정방행렬인 경우, linalg.solve() 사용 가능
np.linalg.solve(X, Y)
````

## 6.2 단변량 회귀분석 2
- 선형 연립방정식
- 환경 : 독립 변수 1개, 데이터 n개

````python
a = np.array([[2,1],[3,1],[4,1],[5,1]])
a
````

````python
# 정방행렬이 아닌 경우, 역행렬 생성 불가
np.linalg.inv(a)
````

````python
# 전치행렬
a.T
````

````python
# a @a.T를 통해 정방행렬을 만들어줌  
a.T @ a
````

````python
# 역행렬 생성을 위한 정방행렬 생성의 경우, 순서 상관없음
a @ a.T
````

````python
np.linalg.inv(a.T @ a)
````

````python
# M = a.T @ a 으로 치환할 경우,M⁻¹@  M로 표현 가능
np.linalg.inv(a.T @ a)@(a.T @ a)
````

````python
x = np.array([32, 63,96, 118, 126, 144, 152.5, 158])
y = np.array([18, 24, 61.5, 49, 52, 105, 130.3, 125])
````

````python
X = np.c_[x, np.ones(len(x))]
Y = y.reshape(-1,1)
print(X)
print(Y)
````

````python
#  X @ wb = Y
# (X @ X.T) @ wb = (Y @ X.T)
# (X @ X.T) @ (np.linalg.inv(X.T @ X)) @ wb = (Y @ X.T) @ (np.linalg.inv(X.T @ X))
# I @ wb = (np.linalg.inv(X.T @ X)) @(X.T @ Y)
wb = np.linalg.inv(X.T @ X) @ (X.T @ Y)
wb
````

````python
# 예측값
h = wb[0] * x + wb[1]
````

````python
print('독립변수 : ', X)
print('종속변수 : ', Y)
print('예측값 : ', h)
````

````python
plt.scatter(x,y)
plt.plot(x, h, c='r') # 선 그래프
````
